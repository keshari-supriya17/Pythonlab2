```python
def calculate_discount(price, discount=10):
    amount = price * discount / 100
    return amount

print(calculate_discount(25000))
print(calculate_discount(3900, 150))
```

# step 7

```python
add = lambda a, b: a + b
subtract = lambda a, b: a - b
multiply = lambda a, b: a * b
divide = lambda a, b: a / b

print("Addition:", add(25, 7))
print("Subtraction:", subtract(36, 2))
print("Multiplication:", multiply(27, 4))
print("Division:", divide(19, 3))
```

```python
employees = [
    {"name": "Ravi", "salary": 50000},
    {"name": "Anita", "salary": 65000},
    {"name": "Kiran", "salary": 45000}
]


sorted_employees = sorted(
    employees,
    key=lambda employee: employee["salary"]
)

for employee in sorted_employees:
    print(employee)
```
